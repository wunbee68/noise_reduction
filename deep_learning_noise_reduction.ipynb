## 2. Noisy·Clean 데이터 불러오기와 캐시

각 분할의 `samples.csv`에 같은 구간의 Noisy·Clean 파일이 한 쌍씩 지정되어 있습니다. **첫 실행**에는 WAV를 읽으면서 표본률과 길이를 확인하고, 학습·검증 파형을 각각 하나의 NumPy 파일로 저장합니다.

**두 번째 실행부터**는 런타임이 재시작되어도 이 파일 두 개를 읽어 RAM에 올립니다. WAV 2,920개를 각각 열어 확인하는 단계는 건너뜁니다. 캐시는 두 분할 합계 약 535 MiB이며, `samples.csv`의 내용이 바뀌면 해당 분할의 캐시를 다시 만듭니다. CSV는 그대로 두고 WAV만 교체했다면 아래 `REBUILD_WAVEFORM_CACHE`를 `True`로 한 번 실행한 뒤 다시 `False`로 돌려두세요.

**1차와 동일한 데이터셋**을 사용합니다. 아래에서는 학습셋의 첫 번째 Noisy·Clean 쌍을 같은 시간·주파수축과 색상 범위로 표시하고, 같은 재생 게인으로 들어 봅니다.

```python
from hashlib import sha256
from io import BytesIO
from tempfile import TemporaryDirectory

REBUILD_WAVEFORM_CACHE = False  # CSV 수정 없이 WAV를 교체했을 때 한 번만 True


def read_wav(path):
    samples, rate = sf.read(path, dtype="float32")
    if rate != SAMPLE_RATE or samples.ndim != 1:
        raise ValueError(f"16 kHz 모노 WAV가 필요합니다: {path}")
    return torch.from_numpy(samples)


def load_waveforms(root, split, manifest, cache_path, sample_count, rebuild=False):
    # [쌍 번호, noisy/clean, 표본 번호] 형태의 연속 배열을 저장합니다.
    shape = (len(manifest), 2, sample_count)
    if cache_path.is_file() and not rebuild:
        try:
            waveforms = np.load(cache_path, allow_pickle=False)
            if waveforms.shape == shape and waveforms.dtype == np.float32:
                print(f"{split}: 저장된 캐시 사용 ({len(manifest):,}쌍)")
                return waveforms
        except (OSError, ValueError, EOFError):
            pass
        print(f"{split}: 읽을 수 없거나 형식이 다른 캐시를 다시 만듭니다.")

    waveforms = np.empty(shape, dtype=np.float32)
    for index, row in enumerate(tqdm(manifest.itertuples(index=False), total=len(manifest),
                                     desc=f"{split}: WAV 읽기")):
        noisy = read_wav(root / row.noisy_path)
        clean = read_wav(root / row.clean_path)
        if noisy.numel() != sample_count or clean.numel() != sample_count:
            raise ValueError(f"파형 길이가 일치하지 않습니다: {row.segment_id}")
        waveforms[index, 0] = noisy.numpy()
        waveforms[index, 1] = clean.numpy()

    cache_path.parent.mkdir(parents=True, exist_ok=True)
    # 저장을 끝낸 파일만 최종 경로로 옮겨 중단된 파일이 캐시로 쓰이지 않게 합니다.
    with TemporaryDirectory(dir=cache_path.parent) as temp_dir:
        temporary_path = Path(temp_dir) / cache_path.name
        np.save(temporary_path, waveforms, allow_pickle=False)
        temporary_path.replace(cache_path)
    print(f"{split}: 캐시 저장 완료 ({len(manifest):,}쌍)")
    return waveforms


class PairedSpeech(Dataset):
    def __init__(self, root, split, cache_dir, rebuild=False):
        manifest_bytes = (root / split / "samples.csv").read_bytes()
        self.manifest = pd.read_csv(BytesIO(manifest_bytes))
        if (self.manifest.empty or self.manifest.segment_id.duplicated().any()
                or not self.manifest["split"].eq(split).all()):
            raise ValueError(f"{split} manifest의 샘플 ID 또는 분할을 확인하세요.")

        lengths = self.manifest["num_samples"]
        sample_count = int(lengths.iloc[0])
        if sample_count <= 0 or not lengths.eq(sample_count).all():
            raise ValueError(f"{split} manifest의 표본 수가 양의 정수로 같아야 합니다.")

        manifest_key = sha256(manifest_bytes).hexdigest()[:20]
        cache_path = cache_dir / f"{split}_v1_{SAMPLE_RATE}_{manifest_key}.npy"
        waveforms = load_waveforms(root, split, self.manifest, cache_path, sample_count, rebuild)
        self.waveforms = torch.from_numpy(waveforms)

    def __len__(self):
        return len(self.manifest)

    def __getitem__(self, index):
        noisy, clean = self.waveforms[index]
        return noisy, clean


print(f"파형 캐시 폴더: {CACHE_DIR}")
train_set = PairedSpeech(DATA_ROOT, "train", CACHE_DIR, REBUILD_WAVEFORM_CACHE)
val_set = PairedSpeech(DATA_ROOT, "validation", CACHE_DIR, REBUILD_WAVEFORM_CACHE)
print(f"학습 {len(train_set):,}쌍  |  검증 {len(val_set):,}쌍  |  파형당 {train_set[0][0].numel():,}개 표본")
```

```python
def comparison_stft(wave):
    """1차 예제와 같은 STFT. Wiener 필터와 비교 그림에 공통으로 사용합니다."""
    wave = np.asarray(wave, dtype=np.float64)
    if wave.size < 512:
        wave = np.pad(wave, (0, 512 - wave.size))
    return signal.stft(wave, fs=SAMPLE_RATE, window="hann", nperseg=512,
                       noverlap=512 - 128, nfft=512, boundary="zeros", padded=True)


def show_comparison(clips, title, noisy_reference=None, playback_peak=0.0):
    # Noisy와 Clean을 첫 행에, 처리 결과를 다음 행에 배치합니다.
    if noisy_reference is None:
        noisy_reference = clips["Noisy input"]
    reference_magnitude = max(np.abs(comparison_stft(noisy_reference)[2]).max(), 1e-12)
    columns = min(2, len(clips))
    rows = (len(clips) + columns - 1) // columns
    fig, axes = plt.subplots(rows, columns, figsize=(12, 3.5 * rows), sharex=True, sharey=True,
                             squeeze=False, constrained_layout=True)
    for ax, (name, wave) in zip(axes.flat, clips.items()):
        frequencies, times, spectrum = comparison_stft(wave)
        magnitude_db = 20 * np.log10(np.maximum(np.abs(spectrum), 1e-12) / reference_magnitude)
        mesh = ax.pcolormesh(times, frequencies / 1000, magnitude_db, shading="auto",
                             cmap="magma", vmin=-80, vmax=0)
        ax.set(title=name, xlabel="Time (s)", xlim=(0, len(wave) / SAMPLE_RATE),
               ylim=(0, SAMPLE_RATE / 2000))
    for ax in axes[:, 0]:
        ax.set_ylabel("Frequency (kHz)")
    for ax in list(axes.flat)[len(clips):]:
        ax.set_visible(False)
    fig.colorbar(mesh, ax=list(axes.flat)[:len(clips)],
                 label="Magnitude (dB relative to noisy STFT peak)")
    fig.suptitle(title)
    plt.show()

    # 기준 음성의 peak를 함께 사용해 epoch 사이의 상대적인 크기를 유지합니다.
    peak = max(playback_peak, max(np.max(np.abs(wave)) for wave in clips.values()))
    playback_gain = 0.99 / max(peak, 1e-8)
    players = []
    for name, wave in clips.items():
        player = Audio(wave * playback_gain, rate=SAMPLE_RATE, normalize=False)._repr_html_()
        players.append(f'<td style="text-align:center;padding:10px"><b>{escape(name)}</b><br>{player}</td>')
    player_rows = ['<tr>' + ''.join(players[start:start + columns]) + '</tr>'
                   for start in range(0, len(players), columns)]
    display(HTML('<table style="width:100%">' + ''.join(player_rows) + '</table>'))
```

## 3. 모델·실험 설정과 예제 CNN

### 3.1. 학습 전에 바꿀 설정

**모델과 실험의 하이퍼파라미터는 아래 셀에서 한 번에 변경합니다.** STFT 크기, CNN 채널 수와 커널 크기, 배치 크기, 학습률, 학습 종료 조건, 비교할 epoch와 검증 샘플 번호를 모았습니다.

- 설정을 바꿨다면 이 셀부터 모델 정의·구조 출력 셀을 다시 실행하고, 4절에서 새로 학습하세요.
- `PREVIEW_INDEX`는 검증셋의 **0부터 시작하는 행 번호**이며, epoch 비교와 Wiener 비교에 함께 사용합니다. 비교 샘플이나 epoch만 바꾸었다면 설정 셀과 해당 비교 셀을 다시 실행하면 됩니다.
- 배포된 가중치는 아래 기본 모델 설정으로 학습되었습니다. 저장된 가중치를 재생할 때는 학습 당시의 STFT·채널·커널 설정을 사용하세요. 모델 설정을 바꾸어 새로 학습하면 학습 기록과 최고 가중치가 `outputs/`에 새로 저장됩니다.


```python
# ── 모델: STFT와 3층 CNN ──
N_FFT = 512                       # STFT 창 길이 (표본 수)
HOP = 128                         # STFT 이동 간격 (표본 수)
HIDDEN_CHANNELS = (64, 64)         # 첫째·둘째 Conv1d의 출력 채널 수
KERNEL_SIZES = (5, 3, 1)           # 세 Conv1d의 시간축 커널 크기 (양의 홀수)

# ── 학습: Adam ──
SEED = 42
BATCH_SIZE = 16
LEARNING_RATE = 0.001
ADAM_BETAS = (0.9, 0.999)          # 1차·2차 모멘트의 이동 평균 계수
WEIGHT_DECAY = 0.0

# ── 수렴: 조기 종료와 학습률 조정 ──
MAX_EPOCHS = 200                   # 최대 실행 길이
MIN_EPOCHS = 30                    # 조기 종료 전 최소 학습 길이
PATIENCE = 20                      # 유의한 PESQ 개선이 없을 때 기다릴 epoch 수
MIN_DELTA = 0.001                  # 조기 종료·학습률 조정의 PESQ 개선 폭
PESQ_EVAL_INTERVAL = 5             # 전체 PESQ 검증 간격 (epoch); 손실 검증은 매 epoch
LR_FACTOR = 0.5                    # PESQ가 정체되면 학습률에 곱할 값
LR_PATIENCE = 5
MIN_LR = 1e-5

# ── 결과 비교: 저장할 epoch와 공통 검증 샘플 ──
PREVIEW_EPOCHS = [10, 20, 40, 80]
PREVIEW_INDEX = 138

assert N_FFT > 0 and 0 < HOP <= N_FFT // 2, "STFT 창 길이와 이동 간격을 확인하세요."
assert len(HIDDEN_CHANNELS) == 2 and all(c > 0 for c in HIDDEN_CHANNELS)
assert len(KERNEL_SIZES) == 3 and all(k > 0 and k % 2 == 1 for k in KERNEL_SIZES)
assert BATCH_SIZE > 0 and 0 < MIN_LR <= LEARNING_RATE
assert len(ADAM_BETAS) == 2 and all(0 <= b < 1 for b in ADAM_BETAS) and WEIGHT_DECAY >= 0
assert 1 <= MIN_EPOCHS <= MAX_EPOCHS and PATIENCE > 0 and MIN_DELTA >= 0
assert PESQ_EVAL_INTERVAL > 0
assert 0 < LR_FACTOR < 1 and LR_PATIENCE >= 0
assert 0 <= PREVIEW_INDEX < len(val_set)
assert PREVIEW_EPOCHS and len(set(PREVIEW_EPOCHS)) == len(PREVIEW_EPOCHS)
assert all(isinstance(e, int) and e >= 1 for e in PREVIEW_EPOCHS)
```

### 3.2. 예제 모델 · CNN으로 마스크 예측하기

STFT는 파형을 짧은 구간으로 나누어 각 구간의 주파수 성분을 구합니다. **기본 설정**에서 512개 표본의 Hann 창을 128개 표본씩 이동하면, 3초 파형 하나는 **257개 주파수 빈 × 376개 시간 프레임**이 됩니다.

CNN의 입력은 잡음 음성 스펙트럼 $Y$의 로그 크기 $\log(1+|Y|)$입니다. 아래 블록도는 **기본 설정**을 나타내며, (a)는 전체 복원 과정, (b)는 코드의 `mask_net`에 해당합니다. 박스 위에는 채널 수, 아래에는 커널 크기와 패딩을 표시했습니다. 출력 마스크 $M$이 0에 가까운 칸은 감쇠되고, 1에 가까운 칸은 입력의 크기를 유지합니다. $M$을 복소수 $Y$에 곱하면 입력 위상을 유지하면서 크기를 바꿀 수 있습니다.

설정을 변경한 모델의 실제 입출력 크기와 파라미터 수는 아래 구조 출력 셀에서 확인하세요.

```python
display(Image(filename=str(DATA_ROOT / "figures" / "cnn_mask.png"), width=1200))
```

`Conv1d`의 입력 순서는 **(배치, 채널, 시간)**입니다. 여기서는 주파수 빈을 채널로 사용합니다. **기본 설정**의 첫 층은 인접한 5개 프레임의 모든 주파수 빈을 함께 보고, 둘째 층은 인접한 3개 프레임의 특징을 결합합니다. 마지막 층은 프레임마다 257개의 마스크 값을 만듭니다. 패딩을 사용하므로 세 층을 지나도 프레임 수는 같습니다. [PyTorch Conv1d](https://docs.pytorch.org/docs/stable/generated/torch.nn.Conv1d.html)

```python
class MaskCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.n_fft, self.hop = N_FFT, HOP
        bins = self.n_fft // 2 + 1
        c1, c2 = HIDDEN_CHANNELS
        k1, k2, k3 = KERNEL_SIZES
        self.mask_net = nn.Sequential(
            nn.Conv1d(bins, c1, kernel_size=k1, padding=k1 // 2),
            nn.ReLU(),
            nn.Conv1d(c1, c2, kernel_size=k2, padding=k2 // 2),
            nn.ReLU(),
            nn.Conv1d(c2, bins, kernel_size=k3, padding=k3 // 2),
            nn.Sigmoid(),
        )
        # 창은 학습 대상이 아니며, model.to(device)를 따라 이동합니다.
        self.register_buffer("window", torch.hann_window(self.n_fft))

    def spectrum(self, wave):
        return torch.stft(wave, n_fft=self.n_fft, hop_length=self.hop, window=self.window,
                          return_complex=True)

    def forward(self, noisy):
        spectrum = self.spectrum(noisy)                # [배치, N_FFT // 2 + 1, 시간]
        features = torch.log1p(spectrum.abs())
        mask = self.mask_net(features)
        return mask * spectrum                        # 복원한 복소수 스펙트럼

    def enhance(self, noisy):
        return torch.istft(self(noisy), n_fft=self.n_fft, hop_length=self.hop,
                           window=self.window, length=noisy.shape[-1])
```

**모델 구조와 파라미터 확인** · 다음 셀은 층의 설정, 입력·출력 텐서 크기, 파라미터 수, 가중치와 편향의 모양을 출력합니다. 입력 표본 하나를 통과시켜 실제 크기를 확인하며, Hann 창은 모델과 함께 저장되는 버퍼입니다.

```python
inspection_model = MaskCNN().eval()
print(inspection_model)
layer_rows = []
with torch.inference_mode():
    inspection_input = train_set[0][0].unsqueeze(0)
    features = torch.log1p(inspection_model.spectrum(inspection_input).abs())
    print(f"Waveform {tuple(inspection_input.shape)} → STFT magnitude {tuple(features.shape)}")
    for name, layer in inspection_model.mask_net.named_children():
        input_shape = tuple(features.shape)
        features = layer(features)
        layer_rows.append({"Layer": f"mask_net.{name} ({type(layer).__name__})",
                           "Input": input_shape, "Output": tuple(features.shape),
                           "Parameters": sum(p.numel() for p in layer.parameters())})
display(pd.DataFrame(layer_rows))
display(pd.DataFrame([{"Tensor": name, "Shape": tuple(parameter.shape),
                       "Parameters": parameter.numel(), "Trainable": parameter.requires_grad}
                      for name, parameter in inspection_model.named_parameters()]))
total_parameters = sum(p.numel() for p in inspection_model.parameters())
trainable_parameters = sum(p.numel() for p in inspection_model.parameters() if p.requires_grad)
print(f"Total parameters {total_parameters:,}  |  Trainable {trainable_parameters:,}")
print(f"Hann window buffer {inspection_model.window.numel():,} values")
```

**학습 진행에 따른 출력 비교** · 3.1절의 `PREVIEW_INDEX`로 검증 샘플을 지정합니다.

기본 설정에서 epoch 비교 그림은 **Noisy·Clean·epoch 10·20·40·80**을 3×2 배열로, Wiener 비교 그림은 **Noisy·Clean·Wiener·CNN**을 2×2 배열로 표시합니다. 각 그림의 모든 패널에 같은 색상 범위와 재생 게인을 적용합니다. 재생 버튼도 그림과 같은 순서입니다.

```python
def wiener_filter(noisy):
    """1차 Wiener 예제의 잡음 추정 및 decision-directed SNR 계산을 사용합니다."""
    noisy = np.asarray(noisy, dtype=np.float64)
    _, times, spectrum = comparison_stft(noisy)
    power = np.abs(spectrum) ** 2

    # 경계 패딩이 들어가지 않은 프레임 중 에너지가 낮은 20%를 선택합니다.
    centers = np.rint(times * SAMPLE_RATE).astype(int)
    full = np.flatnonzero((centers >= 256) & (centers + 256 <= len(noisy)))
    if not len(full):
        full = np.arange(power.shape[1])
    count = max(1, int(np.ceil(0.20 * len(full))))
    energy = power[:, full].mean(axis=0)
    selected = full[np.argsort(energy, kind="stable")[:count]]
    noise = power[:, selected].mean(axis=1)
    noise = np.maximum(uniform_filter1d(noise, size=3, mode="nearest"), 1e-18)

    posterior_snr = power / noise[:, None]
    enhanced = np.empty_like(spectrum)
    previous_clean_power = np.zeros_like(noise)
    for frame in range(spectrum.shape[1]):
        instantaneous = np.maximum(posterior_snr[:, frame] - 1.0, 0.0)
        if frame == 0:
            prior_snr = instantaneous
        else:
            prior_snr = (0.98 * previous_clean_power / noise
                         + (1 - 0.98) * instantaneous)
        gain = np.clip(prior_snr / (1 + prior_snr), 0.05, 1)
        enhanced[:, frame] = gain * spectrum[:, frame]
        previous_clean_power = np.abs(enhanced[:, frame]) ** 2

    _, restored = signal.istft(enhanced, fs=SAMPLE_RATE, window="hann", nperseg=512,
                              noverlap=512 - 128, nfft=512, input_onesided=True, boundary=True)
    return restored[:len(noisy)]


def pesq_wb(clean, estimate):
    score = float(pesq(SAMPLE_RATE, clean, estimate, mode="wb"))
    if not np.isfinite(score) or score < 0:
        raise ValueError(f"유효하지 않은 PESQ-WB 점수: {score}")
    return score


def reference_clips(index):
    noisy, clean = val_set[index]
    return {"Noisy input": noisy.numpy(), "Clean reference": clean.numpy()}


@torch.inference_mode()
def enhance_from_checkpoint(preview_dir, epoch, noisy):
    # 이 함수는 선택한 파일을 해당 epoch의 가중치로 새로 추론합니다.
    snapshot = MaskCNN().to(device)
    snapshot.load_state_dict(torch.load(preview_dir / f"epoch_{epoch:04d}.pt",
                                         map_location=device, weights_only=True))
    snapshot.eval()
    waveform = torch.from_numpy(noisy).unsqueeze(0).to(device)
    return snapshot.enhance(waveform)[0].cpu().numpy()


def show_epoch_comparison(preview_dir, epochs, references, history_table):
    epoch_clips = {f"Epoch {epoch}": enhance_from_checkpoint(preview_dir, epoch, references["Noisy input"])
                   for epoch in epochs}
    clips = {**references, **epoch_clips}
    show_comparison(clips, "Training progress comparison")
    scores = history_table.set_index("epoch").loc[epochs,
        ["train_loss", "validation_loss", "validation_pesq_wb"]].copy()
    scores.columns = ["학습 손실", "검증 손실", "검증 평균 PESQ-WB"]
    scores["예시 파일 PESQ-WB"] = [pesq_wb(references["Clean reference"], wave)
                                      for wave in epoch_clips.values()]
    display(scores.rename_axis("Epoch").round(5))


@torch.inference_mode()
def save_epoch_preview(model, epoch, preview_dir, references):
    model.eval()
    # 가중치도 저장하므로 학습 후 다른 샘플을 선택해 비교할 수 있습니다.
    torch.save(model.state_dict(), preview_dir / f"epoch_{epoch:04d}.pt")
    noisy = torch.from_numpy(references["Noisy input"]).unsqueeze(0).to(device)
    enhanced = model.enhance(noisy)[0].cpu().numpy()
    sf.write(preview_dir / f"epoch_{epoch:04d}.wav", enhanced, SAMPLE_RATE, subtype="FLOAT")
```

## 4. 모델 학습 실행과 검증 PESQ 최고 가중치 선택

**학습 시작 방법:** 1–3절을 실행한 뒤, **4.1절의 함수 정의 셀 → 4.2절의 모델 학습 실행 셀**을 순서대로 실행하세요. 4.2절의 셀을 실행하면 새 모델을 만들고 실제로 역전파와 가중치 갱신을 수행합니다.

모델과 학습 설정은 **3.1절의 설정 셀**에서 변경합니다. 다음 설명은 기본값을 기준으로 합니다.

각 epoch에서는 복원 스펙트럼과 Clean 목표 스펙트럼의 **로그 크기 L1 손실**로 Adam을 갱신하고, 검증 손실을 계산합니다. 전체 검증 PESQ-WB는 기본적으로 `PESQ_EVAL_INTERVAL=5` epoch마다 계산하며, `PREVIEW_EPOCHS`에는 항상 계산합니다. PESQ를 계산한 epoch 중 최고 점수의 가중치를 저장하고, 학습률 조정과 조기 종료도 그 평가 시점에 갱신합니다.

- **속도와 선택 해상도:** PESQ 간격을 늘리면 CPU 평가 시간이 줄지만, 간격 사이에 나온 최고점은 선택되지 않을 수 있습니다. 더 촘촘한 선택이 필요하면 `PESQ_EVAL_INTERVAL=1`로 설정하세요.
- **수렴 판단**은 최소 30 epoch 이후, `MIN_DELTA=0.001` 이상의 개선이 누적 20 epoch 동안 없을 때 조기 종료합니다.
- 검증 PESQ가 정체되면 학습률을 낮춥니다. 최대 실행 길이는 `MAX_EPOCHS=200`이며, 전체 epoch의 손실과 평가된 epoch의 PESQ가 `outputs/training_history.csv`에 저장됩니다.
- `PREVIEW_EPOCHS=[10, 20, 40, 80]`의 가중치와 출력 WAV를 저장합니다. 3.1절에서 샘플을 선택하고 아래 저장된 epoch 비교 셀을 실행하면 Noisy·Clean과 네 시점의 복원 결과를 한 그림에서 비교합니다.

학습 요약과 선택한 epoch는 `outputs/training_summary.json`에 저장됩니다.

### 4.1. 한 epoch의 학습과 검증 함수

`run_epoch`는 noisy 입력의 복원 결과와 사전 캐시된 clean 로그 스펙트럼 목표 사이의 손실로 Adam을 갱신합니다. Clean STFT 캐시는 `.waveform_cache/`에 설정과 manifest별로 저장되며, 학습·검증 합계 약 538 MiB를 추가로 사용합니다. manifest나 STFT 설정이 바뀌면 새 캐시를 만들고, WAV만 교체했다면 `REBUILD_WAVEFORM_CACHE=True`로 다시 생성하세요.

`validation_metrics`는 검증 손실을 매 epoch 계산하고, `PESQ_EVAL_INTERVAL` 간격 및 프리뷰 epoch에서만 전체 PESQ-WB를 계산합니다. 최고 가중치 선택·학습률 조정·조기 종료는 PESQ를 계산한 epoch에서만 갱신됩니다.

```python
def load_target_log_magnitudes(dataset, split, model, rebuild=False):
    manifest_bytes = (DATA_ROOT / split / "samples.csv").read_bytes()
    manifest_key = sha256(manifest_bytes).hexdigest()[:20]
    cache_path = CACHE_DIR / (
        f"{split}_target_logmag_v1_{SAMPLE_RATE}_{model.n_fft}_{model.hop}_{manifest_key}.npy")
    frame_count = dataset.waveforms.shape[-1] // model.hop + 1
    shape = (len(dataset), model.n_fft // 2 + 1, frame_count)

    if cache_path.is_file() and not rebuild:
        try:
            targets = np.load(cache_path, mmap_mode="r", allow_pickle=False)
            if targets.shape == shape and targets.dtype == np.float32:
                print(f"{split}: Clean STFT 캐시 사용 ({len(dataset):,}개)")
            else:
                targets = None
        except (OSError, ValueError, EOFError):
            targets = None
    else:
        targets = None

    if targets is None:
        CACHE_DIR.mkdir(parents=True, exist_ok=True)
        with TemporaryDirectory(dir=CACHE_DIR) as temp_dir:
            temporary_path = Path(temp_dir) / cache_path.name
            target_map = np.lib.format.open_memmap(
                temporary_path, mode="w+", dtype=np.float32, shape=shape)
            model.eval()
            with torch.inference_mode():
                starts = range(0, len(dataset), BATCH_SIZE)
                for start in tqdm(starts, desc=f"{split}: Clean STFT 캐시 생성"):
                    stop = min(start + BATCH_SIZE, len(dataset))
                    clean = dataset.waveforms[start:stop, 1].to(device)
                    target_map[start:stop] = torch.log1p(
                        model.spectrum(clean).abs()).cpu().numpy()
            target_map.flush()
            del target_map
            temporary_path.replace(cache_path)
        targets = np.load(cache_path, mmap_mode="r", allow_pickle=False)
        print(f"{split}: Clean STFT 캐시 저장 ({len(dataset):,}개)")

    with torch.inference_mode():
        sample = dataset.waveforms[0, 1].unsqueeze(0).to(device)
        expected = torch.log1p(model.spectrum(sample).abs())[0].cpu().numpy()
    if not np.allclose(targets[0], expected, rtol=1e-5, atol=1e-6):
        raise RuntimeError(f"{split} Clean STFT 캐시 검증에 실패했습니다.")
    return targets


class CachedTargetDataset(Dataset):
    def __init__(self, paired_dataset, target_log_magnitudes, include_clean=False):
        self.paired_dataset = paired_dataset
        self.target_log_magnitudes = target_log_magnitudes
        self.include_clean = include_clean

    def __len__(self):
        return len(self.paired_dataset)

    def __getitem__(self, index):
        noisy, clean = self.paired_dataset[index]
        target = torch.from_numpy(np.array(
            self.target_log_magnitudes[index], dtype=np.float32, copy=True))
        if self.include_clean:
            return noisy, clean, target
        return noisy, target


def run_epoch(model, loader, optimizer=None):
    training = optimizer is not None
    model.train(training)
    total_loss = 0.0
    with torch.set_grad_enabled(training):
        for noisy, target in loader:
            noisy, target = noisy.to(device), target.to(device)
            estimate = model(noisy)
            loss = nn.functional.l1_loss(torch.log1p(estimate.abs()), target)
            if training:
                optimizer.zero_grad(set_to_none=True)
                loss.backward()
                optimizer.step()
            total_loss += loss.item() * noisy.size(0)
    return total_loss / len(loader.dataset)


@torch.inference_mode()
def validation_metrics(model, loader, compute_pesq=True):
    model.eval()
    total_loss = 0.0
    scores = []
    for noisy, clean, target in loader:
        noisy, target = noisy.to(device), target.to(device)
        estimate = model(noisy)
        loss = nn.functional.l1_loss(torch.log1p(estimate.abs()), target)
        total_loss += loss.item() * noisy.size(0)
        if compute_pesq:
            clean = clean.to(device)
            enhanced = torch.istft(estimate, n_fft=model.n_fft, hop_length=model.hop,
                                   window=model.window, length=noisy.shape[-1]).cpu().numpy()
            for reference, restored in zip(clean.cpu().numpy(), enhanced):
                scores.append(pesq_wb(reference, restored))
    if compute_pesq and len(scores) != len(loader.dataset):
        raise RuntimeError("검증 전체 파일의 PESQ 계산이 필요합니다.")
    mean_pesq = float(np.mean(scores)) if compute_pesq else float("nan")
    return total_loss / len(loader.dataset), mean_pesq
```

### 4.2. ▶ 모델 학습 실행

**아래 코드 셀의 실행 버튼을 누르면 학습이 시작됩니다.** DataLoader, 새 CNN, Adam, 학습률 스케줄러를 준비한 뒤 최대 `MAX_EPOCHS`까지 반복합니다. 진행률에는 매 epoch 학습 손실·검증 손실이, 설정한 PESQ 평가 시점에는 검증 PESQ와 최고 epoch가 표시됩니다.

검증 PESQ가 높아질 때마다 `outputs/best_mask_cnn.pt`를 저장하고, 조기 종료 또는 최대 epoch에 도달하면 **평가된 PESQ 중 최고 가중치를 다시 불러옵니다.** 실행하면 학습 기록과 최고 가중치를 새 실험 결과로 갱신합니다.

```python
# ▶ 모델 학습 실행: 1–3절과 4.1절을 실행한 뒤 이 셀을 실행하세요.
# Clean STFT는 설정별 디스크 캐시에서 읽고, 전체 PESQ는 지정한 간격으로 계산합니다.
torch.manual_seed(SEED)
model = MaskCNN().to(device)
train_target_log_magnitudes = load_target_log_magnitudes(
    train_set, "train", model, rebuild=REBUILD_WAVEFORM_CACHE)
val_target_log_magnitudes = load_target_log_magnitudes(
    val_set, "validation", model, rebuild=REBUILD_WAVEFORM_CACHE)
train_target_set = CachedTargetDataset(train_set, train_target_log_magnitudes)
val_target_set = CachedTargetDataset(val_set, val_target_log_magnitudes, include_clean=True)
train_loader = DataLoader(train_target_set, batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(val_target_set, batch_size=BATCH_SIZE, shuffle=False)
optimizer = torch.optim.Adam(model.parameters(), lr=LEARNING_RATE,
                             betas=ADAM_BETAS, weight_decay=WEIGHT_DECAY)
scheduler_patience = max(
    0, (LR_PATIENCE + PESQ_EVAL_INTERVAL) // PESQ_EVAL_INTERVAL - 1)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=LR_FACTOR, patience=scheduler_patience,
    threshold=MIN_DELTA, threshold_mode="abs", min_lr=MIN_LR,
)
checkpoint = OUTPUT_DIR / "best_mask_cnn.pt"
run_id = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S%fZ")
preview_dir = OUTPUT_DIR / "previews" / run_id
preview_dir.mkdir(parents=True, exist_ok=True)
references = reference_clips(PREVIEW_INDEX)
best_pesq, stopping_reference = -float("inf"), -float("inf")
best_epoch, stale_epochs, last_pesq_epoch = 0, 0, 0
history = []
stop_reason = "max_epochs"
started = perf_counter()

print(f"학습 시작 | 장치 {device} | Train {len(train_set):,}쌍 | Validation {len(val_set):,}쌍 | "
      f"최대 {MAX_EPOCHS} epochs | PESQ 간격 {PESQ_EVAL_INTERVAL}", flush=True)
epoch_progress = tqdm(range(1, MAX_EPOCHS + 1), desc="Model training", unit="epoch")
for epoch in epoch_progress:
    learning_rate = optimizer.param_groups[0]["lr"]
    train_loss = run_epoch(model, train_loader, optimizer)
    compute_pesq = (epoch % PESQ_EVAL_INTERVAL == 0
                    or epoch in PREVIEW_EPOCHS or epoch == MAX_EPOCHS)
    val_loss, val_pesq = validation_metrics(model, val_loader, compute_pesq=compute_pesq)

    if compute_pesq:
        # 최고 가중치와 학습률 스케줄은 전체 PESQ 평가 시점에서 갱신합니다.
        if val_pesq > best_pesq:
            best_pesq, best_epoch = val_pesq, epoch
            temporary_checkpoint = checkpoint.with_suffix(".tmp")
            torch.save(model.state_dict(), temporary_checkpoint)
            temporary_checkpoint.replace(checkpoint)

        if val_pesq >= stopping_reference + MIN_DELTA:
            stopping_reference = val_pesq
            stale_epochs = 0
        else:
            stale_epochs += epoch - last_pesq_epoch
        last_pesq_epoch = epoch
        scheduler.step(val_pesq)

    epoch_progress.set_postfix(train_loss=f"{train_loss:.5f}", val_loss=f"{val_loss:.5f}",
                               val_PESQ=f"{val_pesq:.4f}" if compute_pesq else "skip",
                               best_epoch=best_epoch)
    history.append({"epoch": epoch, "train_loss": train_loss, "validation_loss": val_loss,
                    "validation_pesq_wb": val_pesq, "learning_rate": learning_rate,
                    "best_epoch": best_epoch, "best_pesq_wb": best_pesq,
                    "stale_epochs": stale_epochs, "elapsed_seconds": perf_counter() - started})
    pd.DataFrame(history).to_csv(OUTPUT_DIR / "training_history.csv", index=False)
    if epoch in PREVIEW_EPOCHS:
        print(f"Epoch {epoch:3d} | train {train_loss:.5f} | val loss {val_loss:.5f} | "
              f"val PESQ {val_pesq:.4f} | best {best_pesq:.4f} (epoch {best_epoch}) | "
              f"lr {learning_rate:.2g}", flush=True)
        save_epoch_preview(model, epoch, preview_dir, references)
    if compute_pesq and epoch >= MIN_EPOCHS and stale_epochs >= PATIENCE:
        stop_reason = "early_stopping"
        break

epoch_progress.close()
model.load_state_dict(torch.load(checkpoint, map_location=device, weights_only=True))
model.eval()
training_summary = {
    "run_id": run_id, "status": "complete", "stop_reason": stop_reason,
    "epochs_completed": epoch, "best_epoch": best_epoch, "best_validation_pesq_wb": best_pesq,
    "validation_files": len(val_set), "elapsed_seconds": perf_counter() - started,
    "preview_index": PREVIEW_INDEX, "preview_segment_id": val_set.manifest.iloc[PREVIEW_INDEX].segment_id,
    "preview_directory": str(preview_dir.relative_to(DATA_ROOT)).replace("\\", "/"),
    "config": {"seed": SEED, "batch_size": BATCH_SIZE, "initial_lr": LEARNING_RATE,
               "max_epochs": MAX_EPOCHS, "min_epochs": MIN_EPOCHS, "patience": PATIENCE,
               "min_delta": MIN_DELTA, "pesq_eval_interval": PESQ_EVAL_INTERVAL,
               "lr_patience": LR_PATIENCE, "min_lr": MIN_LR,
               "preview_epochs": PREVIEW_EPOCHS, "sample_rate": SAMPLE_RATE,
               "n_fft": N_FFT, "hop": HOP, "hidden_channels": HIDDEN_CHANNELS,
               "kernel_sizes": KERNEL_SIZES, "adam_betas": ADAM_BETAS,
               "weight_decay": WEIGHT_DECAY, "lr_factor": LR_FACTOR},
    "torch_version": str(torch.__version__), "device": str(device),
}
(OUTPUT_DIR / "training_summary.json").write_text(
    json.dumps(training_summary, ensure_ascii=False, indent=2), encoding="utf-8")
print(f"학습 종료 ({stop_reason}, {epoch} epochs), 최고 가중치 복원 → epoch {best_epoch}, PESQ-WB {best_pesq:.4f}")
```

## 5. 같은 입력에서 Wiener 필터와 딥러닝 비교하기

같은 noisy WAV를 **1차 예제의 Wiener 필터**와 **검증 PESQ가 가장 높았던 CNN**에 각각 입력합니다. **3.1절의 설정 셀**에서 `PREVIEW_INDEX`를 바꾸고, 설정 셀과 샘플 비교 셀을 다시 실행하면 다른 검증 음성을 비교할 수 있습니다. 저장된 가중치는 이 절의 첫 코드 셀에서 불러옵니다.

Wiener 필터는 1차 예제와 같은 Hann 창 512, hop 128을 사용합니다. 에너지가 낮은 완전한 프레임 20%로 잡음 파워를 추정하고, 이전 프레임의 추정치를 반영하는 계수 0.98과 최소 이득 0.05를 적용합니다. 두 방법의 입력은 noisy 파형이며, clean 파형은 결과 비교에 사용합니다.

```python
model = MaskCNN().to(device)
model.load_state_dict(torch.load(OUTPUT_DIR / "best_mask_cnn.pt", map_location=device, weights_only=True))
model.eval()

@torch.inference_mode()
def denoise_waveform(noisy):
    batch = noisy.unsqueeze(0).to(device)  # [1, 표본 수]
    return model.enhance(batch)[0].cpu().numpy()


def denoise_file(path):
    return denoise_waveform(read_wav(path))
```

첫 행의 Noisy·Clean과 둘째 행의 Wiener·딥러닝 출력을 같은 그림에서 비교합니다. 모든 스펙트로그램은 동일한 시간·주파수축과 색상 범위를 사용합니다. 아래 재생 버튼은 그림과 같은 순서이며, 네 음성에 같은 재생 게인을 적용합니다. **3.1절의 설정 셀**에서 `PREVIEW_INDEX`를 바꾸고 실행하면 다른 샘플도 확인할 수 있습니다.

```python
example = val_set.manifest.iloc[PREVIEW_INDEX]
references = reference_clips(PREVIEW_INDEX)
print(f"샘플 {PREVIEW_INDEX} | {example.segment_id} | {example.noise_category} | 목표 SNR {example.target_snr_db:g} dB")
clips = {**references,
         "Wiener filter": wiener_filter(references["Noisy input"]),
         "Deep learning (CNN example)": denoise_waveform(val_set[PREVIEW_INDEX][0])}
show_comparison(clips, "Best validation PESQ checkpoint")
```

## 6. 검증 164개 전체에서 원본과 비교하기

**PESQ-WB**는 16 kHz에서 clean과 비교 대상의 지각적 음질을 평가합니다. **SNR**은 $10\log_{10}(\sum s^2 / \sum(s-\hat{s})^2)$ dB이며, $s$는 clean 파형입니다. 두 지표 모두 값이 클수록 좋습니다. [PESQ 패키지 문서](https://pypi.org/project/pesq/)

각 파일에서 **잡음 입력**과 **CNN 복원**을 동일한 clean 파형에 비교합니다. 재생용 게인은 적용하지 않습니다. 전체 점수는 파일별 점수의 산술평균이고, 개선량은 같은 파일의 복원 점수에서 입력 점수를 뺀 값입니다. 파일별 결과는 `outputs/validation_per_file.csv`, 평균은 `outputs/validation_summary.csv`에서 다시 확인할 수 있습니다.

```python
def snr_db(clean, estimate):
    clean = clean.astype(np.float64)
    error = clean - estimate
    signal_energy = np.sum(clean ** 2)
    error_energy = np.sum(error ** 2)
    if signal_energy == 0:
        raise ValueError("무음 기준 파형의 SNR은 정의되지 않습니다.")
    return np.inf if error_energy == 0 else 10 * np.log10(signal_energy / error_energy)


@torch.inference_mode()
def evaluate_validation(model, dataset):
    model.eval()
    rows = []
    for index, row in enumerate(tqdm(dataset.manifest.itertuples(index=False),
                                     total=len(dataset), desc="Validation PESQ-WB")):
        noisy, clean = dataset[index]
        enhanced = model.enhance(noisy.unsqueeze(0).to(device))[0].cpu().numpy()
        clean, noisy = clean.numpy(), noisy.numpy()
        input_pesq, cnn_pesq = pesq_wb(clean, noisy), pesq_wb(clean, enhanced)
        input_snr, cnn_snr = snr_db(clean, noisy), snr_db(clean, enhanced)
        rows.append({"segment_id": row.segment_id, "noise_category": row.noise_category,
                     "input_pesq_wb": input_pesq, "cnn_pesq_wb": cnn_pesq,
                     "delta_pesq_wb": cnn_pesq - input_pesq,
                     "input_snr_db": input_snr, "cnn_snr_db": cnn_snr,
                     "delta_snr_db": cnn_snr - input_snr})
    return pd.DataFrame(rows)


detail = evaluate_validation(model, val_set)
summary = pd.DataFrame({
    "비교": ["잡음 입력", "CNN 복원", "개선량 (복원 − 입력)"],
    "파일 수": len(detail),
    "평균 PESQ-WB": [detail.input_pesq_wb.mean(), detail.cnn_pesq_wb.mean(), detail.delta_pesq_wb.mean()],
    "평균 SNR (dB)": [detail.input_snr_db.mean(), detail.cnn_snr_db.mean(), detail.delta_snr_db.mean()],
})
detail.to_csv(OUTPUT_DIR / "validation_per_file.csv", index=False, encoding="utf-8-sig")
summary.to_csv(OUTPUT_DIR / "validation_summary.csv", index=False, encoding="utf-8-sig")
display(summary.round(4))
```
